**1. Statistical measures: mean,median,mode,standard deviation and varience with python cose.(with and without function)**

```python
import numpy as np
import statistics
```

```python
#with function
x=[1,2,3,4,5,6,7,8,9,4]
print("Mean: ",np.mean(x))
print("Median: ",np.median(x))
print("Mode: ",statistics.mode(x))
print("Standard deviation: ",np.std(x))
print("Varience: ",np.var(x))
```

```python
#Without function
x=[1,2,3,4,5,4,6,7,8,9]
mean=sum(x)/len(x)
print("Mean: ",mean)
x= sorted(x)
n = len(x)
if n % 2 == 1:
    m = x[n // 2]
    print("Median:",m)
else:
    m = (x[n // 2 - 1] + x[n // 2]) / 2
    print("Median:",m)
freq = {}

for i in x:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

max_freq = max(freq.values())

for key in freq:
    if freq[key] == max_freq:
        mode = key
print("Mode:", mode)
variance = 0

for i in x:
    variance += (i - mean) ** 2

variance = variance / len(x)

print(f"Variance: {variance: .2f}")

# Standard Deviation
std = variance ** 0.5

print("Standard Deviation:", std)
```
