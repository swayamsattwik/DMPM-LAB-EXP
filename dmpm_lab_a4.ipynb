corelation coefficient relation and covarience analysis. Normalization(min-max,z-score,decimal scalling)techniques.

normalization

```python
import pandas as pd
```

```python
df = pd.DataFrame({"Values": [10,20,30]})
df["Min-Max"] = (df["Values"] - df["Values"].min()) / (df["Values"].max() - df["Values"].min())
df["Z-Score"] = (df["Values"] - df["Values"].mean()) / df["Values"].std()
max_abs = df["Values"].abs().max()
j = len(str(int(max_abs)))
df["Decimal_Scaling"] = df["Values"] / (10**j)
```

```python
print(df)
```

correlation and covarience

```python
from scipy.stats import chi2_contingency
```

```python
data=[[20,30],[40,50]]
df = pd.DataFrame(data, index=["Male", "Female"], columns=["Pro_A", "Pro_B"])
print("Contingency table:")
print(df)
```

```python
chi2, p, dof, expected = chi2_contingency(df)
print("\nChi-Square:", chi2)
print("P-value:", p)
print("Degrees of Freedom:", dof)
print("\nExpected Frequencies:")
print(expected)
if p < 0.05:
    print("\nH0 rejected: There is a significant relationship.")
else:
    print("\nH0 acepted: There is no significant relationship.")
```
