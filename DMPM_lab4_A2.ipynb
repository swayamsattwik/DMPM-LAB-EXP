```python
Q1 = df["Close"].quantile(0.25)
Q3 = df["Close"].quantile(0.75)
IQR = Q3 - Q1

outliers = df[(df["Close"] < Q1 - 1.5 * IQR) |
              (df["Close"] > Q3 + 1.5 * IQR)]

print(outliers)
```
