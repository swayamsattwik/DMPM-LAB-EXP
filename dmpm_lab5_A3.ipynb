**3. Introduction to matplotlib and seaborn library. Data visualization.Different ploting techniques and their importance.*

```python
import matplotlib.pyplot as plt
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Simple Line Plot")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = [10, 20, 15, 30, 25]

sns.lineplot(x=[1, 2, 3, 4, 5], y=data)
plt.title("Line Plot using Seaborn")
plt.show()
```

```python
plt.bar(["A", "B", "C"], [20, 35, 25])
plt.show()
```

```python
plt.hist([10, 20, 20, 30, 30, 30, 40])
plt.show()
```

```python
plt.scatter([1, 2, 3, 4], [10, 15, 25, 20])
plt.show()
```

```python
plt.pie([40, 30, 20, 10],
        labels=["A", "B", "C", "D"],
        autopct="%1.1f%%")
plt.show()
```

```python
sns.boxplot(x=[10, 12, 15, 18, 20, 25, 50])
plt.show()
```

```python
sns.heatmap([[1, 2], [3, 4]], annot=True)
plt.show()
```
